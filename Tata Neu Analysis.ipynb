33. write a function to get the region sales dynamically, if we pass any region, that region sales should be displayed?

```python
def get_region_sales(df, region_name):
    region_sales = df.groupby('Region')['Sales'].sum()
    return region_sales.get(region_name, 0)
```

```python
def get_region_sale(df, region):
    region_data = df[df['Region']==region]
    return region_data['Sales'].sum()
```

```python
get_region_sales(df_merged, 'North')
```

34. write a function to get the country sales, profit dynamically, if we pass any country, that country sales, profit should be displayed?

```python
def get_country_profitsales(df, Country):
    profitsales = df.groupby('Country')[['Sales','Profit']].sum()
    return profitsales.loc[Country] 
```

```python
get_country_profitsales(df_merged,'France')
```

35. write a function to get the Category order quantity dynamically, if we pass any Category, that Category order quantity should be displayed.?

```python
def get_Category_OrderQty(df, category_name):
    category_data = df[df['Category']==category_name]
    category_orderqty = category_data['OrderQuantity'].sum()
    return category_orderqty
```

```python
get_Category_OrderQty(df_merged, 'Technology')
```

36. write a function to get the City sales and profit dynamically, if we pass any city, that City sales and profit should be displayed?*

```python
def get_CitySnP(df, city_name):
    city_snp = df.groupby('City')[['Sales','Profit']].sum()
    return city_snp.loc[city_name] if city_name in city_snp.index else False
```

```python
get_CitySnP(df_merged, 'Valence')
```
